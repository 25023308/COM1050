```python
chieu_cao = float(input(" Nhập chiều cao (m): "))
can_nang = float(input(" Nhập cân nặng (kg): "))

BMI = can_nang / (chieu_cao)**2

if BMI >= 30.0 :
    phan_loai = "Béo phì"
elif 25.0 <= BMI < 30.0 :
    phan_loai = "Thừa cân"
elif 18.5 <= BMI < 25.0 :
    phan_loai = "Bình Thường"
else:
    phan_loai = "Thiếu cân"

print(" Chỉ số BMI của bạn là : ", BMI)
print(" Phân loại: ", phan_loai)
```

```python
def in_hoa_don(ds_hang, tong, giam, thanh_toan):
    print("="*50)
    print("{:<20}{:<10} {:<10} {:<10}".format("Tên hàng", "Số lượng", "Đơn giá", "Thành tiền"))
    print("-"*50)
    for hang in ds_hang:
        print("{:<20}{:<10} {:<10.2f} {:<10.2f}".format(hang["ten"], hang["so_luong"], hang["don_gia"], hang["thanh_tien"]))
    print("-"*50)
    print(f"Tổng cộng: {tong:.2f} $")
    print(f"Giảm giá: {giam:.2f} $")
    print(f"Phải trả: {thanh_toan:.2f} $")
    print("="*50)

def main():
    ds_hang = []
    n = int(input(" Nhập số lượng mặt hàng: "))

    for i in range(n):
        print(f"\nMặt hàng {i+1}: ")
        ten = input(" Tên mặt hàng: ")
        so_luong = int(input(" Số lượng: "))
        don_gia = float(input(" Đơn giá ($): "))
        thanh_tien = so_luong * don_gia

        ds_hang.append({"ten": ten, "so_luong": so_luong, "don_gia": don_gia, "thanh_tien": thanh_tien})

    tong = sum(hang["thanh_tien"] for hang in ds_hang)

    if tong >= 100:
        giam = tong * 0.1
    elif tong >= 50:
        giam = tong * 0.05
    else:
        giam = 0

    thanh_toan = tong - giam
    in_hoa_don(ds_hang, tong, giam, thanh_toan)

if __name__ == "__main__":
    main()
```
